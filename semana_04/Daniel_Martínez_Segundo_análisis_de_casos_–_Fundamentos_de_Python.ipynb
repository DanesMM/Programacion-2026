# Problema #1: Calculadora de propina inteligente

## Problema

Se asigna un monto de propina de acuerdo al consumo en la factura.
* Menos de 150.000 CRC un 10%
* Más de 150.000 CRC hasta 225.000 CRC un 7%
* Más de 225.000 CRC un 5% de propina

## Entrada

- Nombre Cliente:
- Número Factura:
- Monto Factura:

## Proceso

- Calcular el porcentaje de propina.

## Salida esperada

- Monto Total de la Propina
- Total de la Factura con Propina Incluida.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal |Nombre: Andrés / Monto: 220000 |Propina: 15400 |Total a pagar: 235400 |
| Caso límite |Nombre:   andres / Monto: 90900 |Propina:45450 |Total a pagar: 954450 |
| Caso inválido |Nombre: Andrés / Monto: 12345a |ERROR | ERROR: Ingrese Monto Válido |

```python
#Entradas
cliente = input("Nombre Cliente: ").upper().strip()
numero_factura = input("Número Factura: ").strip()
try:
  monto_factura = float(input("Monto Factura: "))
except ValueError:
  print("Ingrese un monto válido")

#Proceso
if monto_factura < 150000:
  porcentaje_propina = 0.1
elif monto_factura <= 225000:
  porcentaje_propina = 0.07
else:
  porcentaje_propina = 0.05

#Salida
monto_propina = monto_factura * porcentaje_propina
print(("Monto Propina: CRC"), monto_propina)
total_factura = monto_factura + monto_propina
print(f"{cliente} el total de la factura es: CRC" , total_factura)

```

# Problema #2: Detector de número par/impar

## Problema

Detectar si el número que ingresa la persona es un número par o impar.

## Entrada

- Nombre Usuario:
- Número:

## Proceso

- Identificar un número par e impar.

## Salida esperada

- Su número es:

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal |2 |Par |El número 2 es par |
| Caso límite | 1123456789|Impar |El número 123456789 es impar |
| Caso inválido |123,4 |ERROR |ERROR: Ingrese un número válido |


```python
#Entrada
nombre_usuario = input("Nombre Usuario: ").upper().strip()
try:
  numero = int(input("Número: "))
    #Proceso
  if numero % 2 == 0:
    print(f"{nombre_usuario} el {numero} es par")
  else:
    print(f"{nombre_usuario} el {numero} es impar")
except ValueError:
  print("Ingrese un número válido")
```

# Problema #3: Clasificador de edad

## Problema
Necesitamos clasificar a los jóvenes y adultos de la Iglesia en las Clases de Academia de acuerdo a sus edades.
- De 0 hasta 11 años: Educ. Cristiana Infantil
- De 12 hasta 17 años: C3D
- De 18 hasta 29 años: Vástagos
- De 30 hasta 39 años: Cosechas
- De 40 hasta 55 años: Vencedores
- Mayores de 56 años: Robles Dorados


## Entrada

- Nombre Alumno.
- Edad.

## Proceso

- Identificar a qué clase pertenece la persona de acuerdo a la edad que tiene.

## Salida esperada

- "Nombre alumno" pertenece a la clase "nombre clase"

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal | Daniel / 15 años|Clase C3D | Daniel pertenece a la clase C3D |
| Caso límite |Daniel / 100 años |Robles Dorados |Daniel pertenece a la clase Robles Dorados |
| Caso inválido | Daniel / 18,5| ERROR|ERROR: Ingrese una edad válida |


```python
#Entrada
nombre_alumno = input("Nombre Alumno: ").upper().strip()
try:
  edad = int(input("Edad: "))
except ValueError:
  print("Ingrese una edad válida")

#Proceso:
if edad <12:
  clase = "Educ. Cristiana Infantil"
elif edad >= 12 and edad <= 17:
  clase = "C3D"
elif edad >= 18 and edad <= 29:
  clase = "Vástagos"
elif edad >= 30 and edad <= 39:
  clase = "Cosechas"
elif edad >= 40 and edad <= 55:
  clase = "Vencedores"
elif edad >= 56:
  clase = "Robles Dorados"

#Salida
print(f"{nombre_alumno} pertenece a la clase {clase}")
```

# Problema #4: Validador de compra con descuento

## Problema
El Supermercado tendrá descuentos para todos sus clientes que realicen compras superiores a 150.000 CRC., si la compra es inferior a este monto, tendrá un producto de regalía. Los descuentos serán de la siguiente manera:
- De 150.000 - 250.000: 10% descuento
- De 250.000 - 350.000: 20% descuento
-  Más de 350.000: 30%
- Menos de 150.000: Producto Regalía


## Entrada

- Nombre Cliente:
- Monto de la Compra:

## Proceso

- Asignarle un porcentaje de descuento al cliente de acuerdo al monto de la compra que haya realizado.

## Salida esperada

- "Nombre Cliente", su descuento es de "monto descuento"
- "Nombres Cliente", tiene un producto de regalía.

## Pruebas manuales

| Escenario | Datos de entrada | Resultado esperado | Resultado obtenido |
|-----------|------------------|--------------------|--------------------|
| Caso normal |Juan / 200000 Compra |10% Descuento |Total a pagar: 180000|
| Caso límite |Juan / 149999 Compra |Regalía , Cero Descuento| Regalía |
| Caso inválido |Juan/ trescientos mil |ERROR |ERROR: Ingrese un Monyo válido |


```python
#Entrada
nombre_cliente = input("Nombre Cliente: ").upper().strip()
try:
  monto_compra = float(input("Monto de la Compra: "))
except ValueError:
  print("Ingrese un monto válido")

#Proceso
if monto_compra >= 150000 and monto_compra <= 250000:
  porcentaje_descuento = 0.1
  monto_descuento = monto_compra * porcentaje_descuento
elif monto_compra >= 250000 and monto_compra <= 350000:
  porcentaje_descuento = 0.2
  monto_descuento = monto_compra * porcentaje_descuento
elif monto_compra > 350000:
  porcentaje_descuento = 0.3
  monto_descuento = monto_compra * porcentaje_descuento
else:
  porcentaje_descuento = 0
  monto_descuento = 0
  print(f"{nombre_cliente} TIENE UNA REGALÍA")

#Salida
print("Monto Factura: CRC" , monto_compra)
print("Descuento: CRC" , monto_descuento)
print(f"{nombre_cliente}," , "TOTAL A CANCELAR: CRC" , monto_compra - monto_descuento)
```
